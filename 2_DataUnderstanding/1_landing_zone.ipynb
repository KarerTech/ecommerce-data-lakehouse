## External Data

### EIA Dataset:


```python
import pandas as pd

# the official Excel-URL of the U.S. Energy Information Administration
eia_url = "https://www.eia.gov/dnav/pet/xls/pet_pri_gnd_dcus_nus_w.xls"

# weekly average prices
# skip the first 2 rows as they contain metadata
df_eia = pd.read_excel(eia_url, sheet_name="Data 1", skiprows=2)

# Rename columns
df_eia = df_eia.rename(columns={
    df_eia.columns[0]: "date",
    df_eia.columns[1]: "gasoline_all_grades",
    df_eia.columns[3]: "gasoline_regular",
    df_eia.columns[12]: "diesel"
})

# Convert the 'date' column to datetime format and filter the DataFrame for the years 2019 to 2024
df_eia["date"] = pd.to_datetime(df_eia["date"])
df_us_prices = df_eia[(df_eia["date"] >= "2019-01-01") & (df_eia["date"] <= "2024-12-31")].copy()

print(f"Data loaded successfully! Time period: 2019-2024: {len(df_us_prices)}")
```

```python
# Only keep the relevant columns for gasoline and diesel prices 
gas_prices_us = df_us_prices[["date", "gasoline_regular", "diesel"]].copy()
```

### US Holidays

#### Create our own list with the Python holidays library

```python

# A selection of important US holidays + shopping events
# 2019-2024 
# save as: us_holidays.csv

from datetime import datetime, timedelta
import pandas as pd
import holidays


holiday_data= []

# 1. Federal Holidays
for year in range(2019, 2025):
    us_hols = holidays.US(years=year)
    for date, name in us_hols.items():
        holiday_data.append({'date': date, 'holiday_name': name, 'type': 'Federal'})

# Helper functions to calculate the dates of certain holidays
def easter_date(year):
    a = year % 19
    b = year // 100
    c = year % 100
    d = b // 4
    e = b % 4
    f = (b + 8) // 25
    g = (b - f + 1) // 3
    h = (19 * a + b - d - g + 15) % 30
    i = c // 4
    k = c % 4
    l = (32 + 2 * e + 2 * i - h - k) % 7
    m = (a + 11 * h + 22 * l) // 451
    month = (h + l - 7 * m + 114) // 31
    day = ((h + l - 7 * m + 114) % 31) + 1
    return datetime(year, month, day).date()

def mothers_day(year):
    may_1st = datetime(year, 5, 1)
    first_sunday = may_1st + timedelta(days=(6 - may_1st.weekday()))
    return (first_sunday + timedelta(days=7)).date()

def fathers_day(year):
    june_1st = datetime(year, 6, 1)
    first_sunday = june_1st + timedelta(days=(6 - june_1st.weekday()))
    return (first_sunday + timedelta(days=14)).date()

def thanksgiving(year):
    nov_1st = datetime(year, 11, 1)
    # Find the first Thursday in November
    days_ahead = (3 - nov_1st.weekday()) % 7
    first_thursday = nov_1st + timedelta(days=days_ahead)
    # 4. Thursday = 3. + 21 Days
    return (first_thursday + timedelta(days=21)).date()

# 2. Cultural Holidays + Shopping Events
for year in range(2019, 2025):
    # Cultural
    holiday_data.append({'date': datetime(year, 2, 14).date(), 'holiday_name': 'Valentine\'s Day', 'type': 'Cultural'})
    holiday_data.append({'date': datetime(year, 3, 17).date(), 'holiday_name': 'St. Patrick\'s Day', 'type': 'Cultural'})
    holiday_data.append({'date': easter_date(year), 'holiday_name': 'Easter Sunday', 'type': 'Cultural'})
    holiday_data.append({'date': mothers_day(year), 'holiday_name': 'Mother\'s Day', 'type': 'Cultural'})
    holiday_data.append({'date': fathers_day(year), 'holiday_name': 'Father\'s Day', 'type': 'Cultural'})
    holiday_data.append({'date': datetime(year, 10, 31).date(), 'holiday_name': 'Halloween', 'type': 'Cultural'})
    
    # Shopping Events
    thanksgiving_date = thanksgiving(year)
    holiday_data.append({'date': thanksgiving_date - timedelta(days=1), 'holiday_name': 'Day Before Thanksgiving', 'type': 'Shopping'})
    holiday_data.append({'date': thanksgiving_date + timedelta(days=1), 'holiday_name': 'Black Friday', 'type': 'Shopping'})
    holiday_data.append({'date': thanksgiving_date + timedelta(days=4), 'holiday_name': 'Cyber Monday', 'type': 'Shopping'})
    holiday_data.append({'date': datetime(year, 7, 15).date(), 'holiday_name': 'Amazon Prime Day', 'type': 'Shopping'})

# Create DataFrame, sort by date, and remove duplicates
us_holidays = pd.DataFrame(holiday_data)
us_holidays['date'] = pd.to_datetime(us_holidays['date'])
us_holidays = us_holidays.sort_values('date').drop_duplicates(subset=['date', 'holiday_name']).reset_index(drop=True)

# Save the final DataFrame to a CSV file
us_holidays.to_csv('us_holidays.csv', index=False)

us_holidays.head()
```
