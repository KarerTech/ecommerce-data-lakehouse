## 1. dim_users

```python
import pandas as pd
from sqlalchemy import create_engine
import time
import gc

print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

users = pd.read_sql("SELECT * FROM iceberg.warehouse.users", con=engine)

dim_users = users[['id', 'age', 'gender', 'country', 'state', 'city', 'traffic_source', 'created_at']].copy()
dim_users.rename(columns={'id': 'user_id', 'created_at': 'registered_at'}, inplace=True)

start_time = time.time()

print(f"Write {len(dim_users):,} rows...")
dim_users.to_sql(
    name='dim_users', 
    con=engine, 
    schema='datamart',
    index=False, 
    chunksize=5000,    
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")
```

## 2. dim_products

```python
import pandas as pd
from sqlalchemy import create_engine
import time
import gc

print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

products = pd.read_sql("SELECT * FROM iceberg.warehouse.products", con=engine)
dc = pd.read_sql("SELECT * FROM iceberg.warehouse.distribution_centers", con=engine)

dim_products = pd.merge(products, dc, left_on='distribution_center_id', right_on='id', how='left')

dim_products = dim_products[['id_x', 'name', 'category', 'brand', 'department', 'retail_price', 'id_y', 'city', 'latitude', 'longitude']].copy()
dim_products.columns = ['product_id', 'product_name', 'category', 'brand', 'department', 'retail_price', 'dc_id', 'dc_city', 'dc_latitude', 'dc_longitude']

start_time = time.time()

print(f"Write {len(dim_products):,} rows...")
dim_products.to_sql(
    name='dim_products', 
    con=engine, 
    schema='datamart',
    index=False, 
    chunksize=2000,    
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")
```

## 3. dim_date

```python
import pandas as pd
from sqlalchemy import create_engine
import time
import gc

print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

us_holidays = pd.read_sql("SELECT * FROM iceberg.warehouse.us_holidays", con=engine)
gas = pd.read_sql("SELECT * FROM iceberg.warehouse.us_gas_prices", con=engine)

df_date = pd.DataFrame({"date_key": pd.date_range(start="2019-01-01", end="2024-12-31")})
df_date["year"] = df_date["date_key"].dt.year
df_date["quarter"] = df_date["date_key"].dt.quarter
df_date["month"] = df_date["date_key"].dt.month
df_date["week"] = df_date["date_key"].dt.isocalendar().week.astype(int)
df_date["day"] = df_date["date_key"].dt.day
df_date["day_of_week"] = df_date["date_key"].dt.dayofweek + 1
df_date["is_weekend"] = df_date["day_of_week"].isin([6, 7])
df_date["date_str"] = df_date["date_key"].dt.strftime("%Y-%m-%d")

us_holidays["date_str"] = pd.to_datetime(us_holidays["date"]).dt.strftime("%Y-%m-%d")
us_holidays = us_holidays.drop_duplicates(subset=["date_str"])

gas["date_str"] = pd.to_datetime(gas["date"]).dt.strftime("%Y-%m-%d")
gas = gas.sort_values("date_str")

df_date = pd.merge(df_date, gas[["date_str", "gasoline_regular", "diesel"]], on="date_str", how="left")
df_date["gasoline_regular"] = df_date["gasoline_regular"].ffill().bfill()
df_date["diesel"] = df_date["diesel"].ffill().bfill()

df_date = pd.merge(df_date, us_holidays[["date_str", "holiday_name", "type"]], on="date_str", how="left")

df_date["holiday_name"] = df_date["holiday_name"].fillna("")
df_date["is_public_holiday"] = df_date["type"].fillna("") == "Federal"

dim_date = df_date[["date_str", "year", "quarter", "month", "week", "day", "day_of_week", "is_weekend", "holiday_name", "is_public_holiday", "gasoline_regular", "diesel"]].copy()
dim_date.columns = [
    'date_key', 'year', 'quarter', 'month', 'week', 'day', 'day_of_week', 
    'is_weekend', 'holiday_name', 'is_public_holiday', 
    'us_gas_price_regular_avg', 'us_gas_price_diesel_avg'
]
start_time = time.time()

print(f"Write {len(dim_date):,} rows...")
dim_date.to_sql(
    name='dim_date', 
    con=engine, 
    schema='datamart',
    index=False, 
    chunksize=2000,    
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")
```

## 4. fact_sales

```python
import pandas as pd
from sqlalchemy import create_engine
import time
import gc

print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

oi = pd.read_sql("SELECT * FROM iceberg.warehouse.order_items", con=engine)
ii = pd.read_sql("SELECT id, cost FROM iceberg.warehouse.inventory_items", con=engine)

fact_sales = pd.merge(oi, ii, left_on='inventory_item_id', right_on='id', how='left')

fact_sales['date_key'] = pd.to_datetime(fact_sales['created_at']).dt.strftime('%Y-%m-%d')
fact_sales['inventory_cost'] = fact_sales['cost']
fact_sales['gross_margin'] = fact_sales['sale_price'] - fact_sales['inventory_cost']
fact_sales['is_returned'] = fact_sales['status'].apply(lambda x: 1 if x == 'Returned' else 0)

fact_sales = fact_sales[['id_x', 'order_id', 'user_id', 'product_id', 'date_key', 'created_at', 'status', 'sale_price', 'inventory_cost', 'gross_margin', 'is_returned']].copy()
fact_sales.columns = ['order_item_id', 'order_id', 'user_id', 'product_id', 'date_key', 'created_at', 'status', 'sale_price', 'inventory_cost', 'gross_margin', 'is_returned']

start_time = time.time()

print(f"Write {len(fact_sales):,} rows...")
fact_sales.to_sql(
    name='fact_sales', 
    con=engine, 
    schema='datamart',
    index=False, 
    chunksize=5000,    
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")

del oi
del ii
del fact_sales
gc.collect()
```
