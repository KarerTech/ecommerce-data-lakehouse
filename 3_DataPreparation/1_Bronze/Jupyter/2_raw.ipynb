## Looker Dataset:

## 1. distribution_centers

```python
import pandas as pd

df = pd.read_csv('../LookerDataset/distribution_centers.csv')


```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import pandas as pd
from sqlalchemy import create_engine
import time


print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

print("Read CSV...")
start_time = time.time()
df = pd.read_csv('../LookerDataset/distribution_centers.csv')

print(f"Write {len(df):,} rows...")
df.to_sql(
    name='distribution_centers', 
    con=engine, 
    schema='raw',
    index=False, 
    chunksize=3500,   
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")
```

```python
import pandas as pd
from sqlalchemy import create_engine
import time

print("try connect to trino...")
engine = create_engine("trino://admin@trino:8080/iceberg")

print("Read CSV...")
start_time = time.time()
df = pd.read_csv('../LookerDataset/events.csv')

df['user_id'] = df['user_id'].astype('Int64')

df['created_at'] = pd.to_datetime(df['created_at'], format='mixed')

print(f"Write {len(df):,} rows...")
df.to_sql(
    name='events', 
    con=engine, 
    schema='raw',
    index=False, 
    chunksize=3500,   
    method='multi'    
)

end_time = time.time()
print("---")
print(f" Done! in: {round(end_time - start_time, 2)} seconds.")
```
